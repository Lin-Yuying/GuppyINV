```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. LD decay

```python
df_QuH = pd.read_csv("/Volumes/LaCie/QuH.LDdecy.stat.gz", header='infer', delimiter="\t")
df_QuL = pd.read_csv("/Volumes/LaCie/QuL.LDdecy.stat.gz", header='infer', delimiter="\t")
df_ArH = pd.read_csv("/Volumes/LaCie/ArH.LDdecy.stat.gz", header='infer', delimiter="\t")
df_ArL = pd.read_csv("/Volumes/LaCie/ArL.LDdecy.stat.gz", header='infer', delimiter="\t")
df_YaH = pd.read_csv("/Volumes/LaCie/YaH.LDdecy.stat.gz", header='infer', delimiter="\t")
df_YaL = pd.read_csv("/Volumes/LaCie/YaL.LDdecy.stat.gz", header='infer', delimiter="\t")
# df_Quare = pd.read_csv("/Users/rainlam/Quare_LDdecay.stat", header='infer', delimiter="\t")
# df_Yarra = pd.read_csv("/Users/rainlam/Yarra_LDdecay.stat", header='infer', delimiter="\t")
```

```python
df_ArH.columns
colors ={'ArH':'orangered', 'ArL':'darkred', 'QuH':'olivedrab', 'QuL':'darkgreen', 'YaH':'cornflowerblue', 'YaL':'darkblue'}
```

```python
plt.figure(figsize=(10,8), dpi = 300)
plt.plot(df_ArH['#Dist']/1e3, df_ArH['Mean_r^2'], label='ArH', color=colors["ArH"])
plt.plot(df_ArL['#Dist']/1e3, df_ArL['Mean_r^2'], label='ArL', color=colors["ArL"])
plt.plot(df_QuH['#Dist']/1e3, df_QuH['Mean_r^2'], label='QuH', color=colors["QuH"])
plt.plot(df_QuL['#Dist']/1e3, df_QuL['Mean_r^2'], label='QuL', color=colors["QuL"])
plt.plot(df_YaH['#Dist']/1e3, df_YaH['Mean_r^2'], label='YaH', color=colors["YaH"])
plt.plot(df_YaL['#Dist']/1e3, df_YaL['Mean_r^2'], label='YaL', color=colors["YaL"])
plt.ylabel("Mean r$^2$", fontsize=16, weight="bold")
plt.xlabel("Distance (kb)", fontsize=16, weight="bold")
# plt.xticks(list(range(0,11000, 1000)),list(range(0,11,1)), fontsize=12)
# plt.xlim(-1000,10000)
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
# plt.legend(title="Populations")
plt.legend(title="Populations", loc="upper right", 
           prop={'size': 'x-large'}, title_fontsize="x-large")


plt.savefig("LDdecay_6pop.pdf", format="pdf")
```
